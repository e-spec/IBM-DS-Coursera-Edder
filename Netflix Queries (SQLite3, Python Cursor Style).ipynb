```python
# Find the top 10 countries with the most titles
cur = conn.cursor()
cur.execute("""
SELECT country, COUNT(*) AS total
FROM netflix_titles
WHERE country IS NOT NULL
GROUP BY country
ORDER BY total DESC
LIMIT 10;
""")
for row in cur.fetchall():
    print(row)
```

```python
# Show the most common ratings (like TV-MA, PG, etc.)
cur = conn.cursor()
cur.execute("""
SELECT rating, COUNT(*) AS total
FROM netflix_titles
GROUP BY rating
ORDER BY total DESC;
""")
for rating, total in cur.fetchall():
    print(f"{rating}: {total}")
```

```python
# List titles released after 2020, newest first
cur = conn.cursor()
cur.execute("""
SELECT title, release_year
FROM netflix_titles
WHERE release_year > 2020
ORDER BY release_year DESC;
""")
rows = cur.fetchall()
for title, year in rows:
    print(f"{year} - {title}")
```

```python
# Search for titles where a specific actor appears in the cast
actor = "Bruce Willis"
cur = conn.cursor()
cur.execute(f"""
SELECT title, "cast"
FROM netflix_titles
WHERE "cast" LIKE '%{actor}%';
""")
for title, cast in cur.fetchall():
    print(f"{title} → {cast}")
```

```python
# Find Mexican movies with the highest rating
query = """
SELECT title, rating, release_year
FROM netflix_titles
WHERE type = 'Movie'
  AND country LIKE '%Mexico%'
ORDER BY rating DESC;
"""
cur = conn.cursor()
cur.execute(query)
for row in cur.fetchall():
    print(row)
```

```python
# Find Mexican movies released in 2023 with the highest rating
query = """
SELECT title, rating, release_year, country
FROM netflix_titles
WHERE type = 'Movie'
  AND country LIKE '%Mexico%'
  AND release_year = 2021
ORDER BY rating DESC;
"""
cur = conn.cursor()
cur.execute(query)
for row in cur.fetchall():
    print(row)
```
