```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
%matplotlib inline

# Sample Data
np.random.seed(42)
X = np.random.rand(100, 2)  # 100 points in 2D

# KMeans Clustering
k_means4 = KMeans(init="k-means++", n_clusters=8, n_init=12, random_state=42)
k_means4.fit(X)

# Plotting
fig, ax = plt.subplots(figsize=(6, 4))
colors = plt.cm.tab10(np.linspace(0, 1, len(set(k_means4.labels_))))

for k, col in zip(range(len(k_means4.cluster_centers_)), colors):
    my_members = (k_means4.labels_ == k)
    cluster_center = k_means4.cluster_centers_[k]

    # Scatter plot for cluster points
    ax.scatter(X[my_members, 0], X[my_members, 1], color=col, marker='.', s=50, label=f'Cluster {k}')
    
    # Plot cluster centers
    ax.scatter(cluster_center[0], cluster_center[1], color=col, edgecolors='k', marker='o', s=150, linewidth=2)

ax.set_title('K-Means Clustering')
ax.legend()
plt.show()
```
